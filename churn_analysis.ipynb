# Feature Engineering

```python
# Create new column
df['Tenure_Group'] = pd.cut(
    df['Tenure'], 
    bins=[-1, 6, 12, 24, 100], 
    labels=['0-6 Months', '6-12 Months', '1-2 Years', '2+ Years']
)
```

```python
#Distance group
df['Distance_Group'] = pd.cut(df['WarehouseToHome'], bins=[-1, 10, 20, 150], 
                              labels=['Near (<10km)', 'Medium (10-20km)', 'Far (>20km)'])
```

```python
# Recency Group (Day Since Last Order)
df['Recency_Group'] = pd.cut(df['DaySinceLastOrder'], bins=[-1, 3, 7, 15, 100], 
                             labels=['0-3 Days', '4-7 Days', '8-15 Days', '15+ Days'])
```
